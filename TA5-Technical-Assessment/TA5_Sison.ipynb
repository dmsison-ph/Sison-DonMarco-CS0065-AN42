## Part I: TextBlob Sentiment Analysis

This program accepts ten sentences, determines each sentence's polarity, and summarizes the number of positive, neutral, and negative sentences.

```python
from textblob import TextBlob

def main():
    counts = {"Positive": 0, "Neutral": 0, "Negative": 0}

    print("==========================================")
    print("Sentiment Analysis Summarizer by Don Sison")
    print("==========================================")

    for number in range(1, 11):
        sentence = input(f"\nEnter sentence {number}: ")
        polarity, label = analyze_sentence(sentence)
        counts[label] += 1
        print(f"Polarity: {polarity:.2f}")
        print(f"Sentiment: {label}")
    print_summary(counts)

def analyze_sentence(sentence):
    polarity = TextBlob(sentence).sentiment.polarity

    if polarity < 0:
        label = "Negative"
    elif polarity == 0:
        label = "Neutral"
    else:
        label = "Positive"

    return polarity, label

def print_summary(counts):
    print()
    print("==========================================")
    print("          Results of the analysis")
    print("==========================================")
    print(f"Your positive sentence count is {counts['Positive']}")
    print(f"Your neutral sentence count is {counts['Neutral']}")
    print(f"Your negative sentence count is {counts['Negative']}")
    print(f"Total sentences: {sum(counts.values())}")
    print("==========================================")

main()
```

## Part II: Challenge — Happiness and Sadness Analyzer

This program uses the public NLTK Opinion Lexicon to count happy and sad words. It classifies the sentence as positive, negative, or neutral by comparing both totals.

```python
import re
from nltk.corpus import opinion_lexicon

happy_words = set(opinion_lexicon.positive())
sad_words = set(opinion_lexicon.negative())

def main():
    print("===========================================")
    print("Happiness and Sadness Analyzer by Don Sison")
    print("===========================================")

    for number in range(1, 4):
        print()
        sentence = input(f"Enter sentence {number} of 3: ")

        happy_count = number_happy(sentence)
        sad_count = number_sad(sentence)

        print()
        print(f"========== Sentence {number} Result ==========")
        print(f"Happy words: {happy_count}")
        print(f"Sad words:   {sad_count}")

        if happy_count > sad_count:
            print("Analysis: Positive")
            print("Tone: Happy")
        elif sad_count > happy_count:
            print("Analysis: Negative")
            print("Tone: Sad")
        else:
            print("Analysis: Neutral")
            print("Tone: Neutral")

    print()
    print("===========================================")
    print("       All 3 sentences analyzed")
    print("===========================================")

def get_words(sentence):
    return re.findall(r"[a-zA-Z]+(?:[-'][a-zA-Z]+)*", sentence.lower())

def number_happy(sentence):
    return sum(1 for word in get_words(sentence) if word in happy_words)

def number_sad(sentence):
    return sum(1 for word in get_words(sentence) if word in sad_words)

main()
```
